```python
# Performing EDA
print("Data Information \n", df.info())
print("Data Shape : ", df.shape)
print("Data Description : \n", df.describe())
```

```python
y = df['Close']
X = df[['Open', 'High', 'Low', 'Adj Close', 'Volume']]
```

```python
print("Input Data", X.columns)
print("Prediction Data: Close" )
```

```python
# applying train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size = 0.2, random_state = 24
)
```

```python
print("Training Data Shape : ", X_train.shape)
print("Testing Data Shape : ", X_test.shape)
```

```python
# Scaling the data
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
# Applying Gradient Descent
import numpy as np
def linear_gd(X, y, lr=0.01, epochs=1000, alpha=0.1):
    n, m = X.shape
    w = np.zeros(m)
    b = 0

    for i in range(epochs):
        pred = np.dot(X, w) + b
        loss = pred - y

        dw = (2/n) * np.dot(X.T, loss)
        db = (2/n) * np.sum(loss)

        w -= lr * dw
        b -= lr * db

    return w,b

def ridge_gd(X, y, lr=0.01, epochs=1000, alpha=0.1):
    n, m = X.shape
    w = np.zeros(m)
    b = 0

    for i in range(epochs):
        pred = np.dot(X, w) + b
        loss = pred - y

        dw = (2/n) * np.dot(X.T, loss) +2 * alpha * w
        db = (2/n) * np.sum(loss)

        w -= lr * dw
        b -= lr * db

    return w,b

def lasso_gd(X, y, lr=0.01, epochs=1000, alpha=0.1):
    n, m = X.shape
    w = np.zeros(m)
    b = 0

    for i in range(epochs):
        pred = np.dot(X, w) + b
        loss = pred - y

        dw = (2/n) * np.dot(X.T, loss) + alpha *np.sign(w)
        db = (2/n) * np.sum(loss)

        w -= lr * dw
        b -= lr * db

    return w,b
```

```python
w_linear, b_linear = linear_gd(X_train_scaled, y_train.values)
pred_linear = np.dot(X_test_scaled, w_linear) + b_linear
print("Linear Regression Successfull")
```

```python
w_ridge, b_ridge = ridge_gd(X_train_scaled, y_train.values, alpha=0.1)
pred_ridge = np.dot(X_test_scaled, w_ridge) + b_ridge
print("Ridge Regression Successfull")
```

```python
w_lasso, b_lasso = lasso_gd(X_train_scaled, y_train.values, alpha=0.1)
pred_lasso = np.dot(X_test_scaled, w_lasso) + b_lasso
print("Lasso Regression Successfull")
```

```python
def evaluate(name, actual, predicted):
    mae = mean_absolute_error(actual, predicted)
    rmse = np.sqrt(mean_squared_error(actual, predicted))
    r2 = r2_score(actual, predicted)


    print("Regression Type : ", name)
    print("MAE : ", mae)
    print("RMSE : ", rmse)
    print("R2 :", r2)
    print("\n")

    return name, mae, rmse, r2
    
```

```python
linear_result = evaluate("Linear Regression", y_test, pred_linear)
ridge_result = evaluate("Ridge Regression", y_test, pred_ridge)
lasso_result = evaluate("Lasso Regression", y_test, pred_lasso)
```

```python
train_linear = np.dot(X_train_scaled, w_linear) + b_linear
train_ridge = np.dot(X_train_scaled, w_ridge) + b_ridge
train_lasso = np.dot(X_train_scaled, w_lasso) + b_lasso

print("Comparing the training; the testing accuracy")
print("Linear Regression")
print("Training R2 : ", r2_score(y_train, train_linear))
print("testing R2 :", r2_score(y_test, pred_linear))

print("Ridge Regression")
print("Training R2 : ", r2_score(y_train, train_ridge))
print("testing R2 :", r2_score(y_test, pred_ridge))

print("Lasso Regression")
print("Training R2 : ", r2_score(y_train, train_lasso))
print("testing R2 :", r2_score(y_test, pred_lasso))
```

```python
kf = KFold(n_splits = 5, shuffle = True, random_state= 21)

scores = []

for train_idx, val_idx in kf.split(X):
    X_tr = X.iloc[train_idx]
    X_val = X.iloc[val_idx]

    y_tr = y.iloc[train_idx]
    y_val = y.iloc[val_idx]

    fold_scaler = StandardScaler()

    X_tr = fold_scaler.fit_transform(X_tr)
    X_val = fold_scaler.transform(X_val)

    w, b = linear_gd(X_tr, y_tr.values)
    pred = np.dot(X_val, w) + b

    score = r2_score(y_val, pred)
    scores.append(score)

print("R2 Scores using cross-validation")
print(scores)
```
