# Importing all the required libraries

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler, MinMaxScaler
```

# Loading the dataset

```python
df = pd.read_csv('prime_data.csv', encoding='latin-1')
df.head()
```

# Extracting the infromation about the dataset

```python
print(df.info())
print(df.describe())
print(df.isnull())
print(df.shape)
```

# Performing the pre-processing on the dataset

```python
# handling missing values
df['Name of the show'].isnull().sum()
columns = df.columns

for cols in columns:
    if df[cols].isnull().sum() > 1:
        df[cols].fillna(method='ffill', inplace=True)
```

```python
print("Count of null values : \n",df.isnull().sum())
```

```python
# removing duplicates
df.drop_duplicates(inplace=True)
print("Checking the new number of duplicates : ",df.duplicated().sum())
```

```python
# applying encoding techniques

print(df.dtypes)
le = LabelEncoder()

for cols in columns:
    if df[cols].dtype == "object":
        df["encoded"+cols] = le.fit_transform(df[cols])

df.head(2)
print("\nAfter encoding : \n", df.dtypes)
```

```python
# removing the unrequired columns

for cols in columns:
    if df[cols].dtype =='object':
        df.drop([cols], axis=1, inplace=True)
```

```python
df.head(2)
```

```python
# applying normalisation, standardisation

normalisation = MinMaxScaler()
standardisation = StandardScaler()

df_normalised = normalisation.fit_transform(df[['Year of release', 'No of seasons available']])
df_standardised = standardisation.fit_transform(df)

print("Normalisation \n",df_normalised)
print("Standardisation \n",df_standardised)
```
